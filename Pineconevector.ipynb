```python
from langchain_core.chains import create_retrieval_chain
from langchain_core.chains.combine_documents import create_stuff_documents_chain
```

```python
from langchain_community.document_loaders import PyPDFDirectoryLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, OpenAI
from langchain_community.vectorstores import Pinecone

from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.documents import Document

import os
```

```python
loader = PyPDFDirectoryLoader("pdfs")
data = loader.load()
```

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=20)
```

```python
text_chunks = text_splitter.split_documents(data)
```

```python
import os
os.environ['OPENAI_API_KEY']
```

```python
embedding = OpenAIEmbeddings()
```

```python
len(embedding.aembed_query("How are you"))
```

```python
PINECONE_API_KEY = os.environ.get('PINECONE_API_KEY','key available in pinecone website')
PINECONE_API_ENV = os.environ.get('PINECONE_API_ENV', 'env name available in pinecone website')
```

```python
import pinecone
```

```python
pinecone.init(api_key PINECONE_API_KEY,Environment PINECONE_API_ENV)
```

```python
index_name = "index_name"
```

```python
index = pinecone.Index("testing")
```

Create Embedding for each of the Text Chunk

```python
docsearch = pinecone.from_texts([t.page_content for t in text_chunks], embedding, index_name=index_name)
```

```python
docsearch
```

```python
query = "YOLov7 outperforms which models"
```

```python
docs = docsearch.similarity_search(query)
```
